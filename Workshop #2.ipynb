* Exercise 2: Write a function that receives the dataset and replace Famle with F and Male with M (please try to write it yourself before checking the answer in the next cell)

```python
# exercise 2 solution
def encode_sex(data):
    data.reset_index(drop=True, inplace=True) # reindexing the dataset in case the dataset index is corrupted
    rows=data.shape[0]
    a=data.loc[:,'sex']
    for i in range(rows):
        if a[i]=="Male":
            data.loc[i,"sex"]="M"
        elif a[i]=="Female":
            data.loc[i,"sex"]="F"
    return data
```

```python
############# WRITE YOUR CODE IN THIS CELL (IF APPLICABLE)  ####################
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

# Step 1: Identify categorical and numerical columns automatically
categorical_columns = adultFrame_sample_cleaned.select_dtypes(
    include='object'
).columns.tolist()
numerical_columns = adultFrame_sample_cleaned.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Categorical columns:", categorical_columns)
print("Numerical columns  :", numerical_columns)
```

```python
# Step 2: Encode categorical columns using LabelEncoder
def encode_categorical_columns(df):
    label_encoders = {}
    for col in df.select_dtypes(include='object').columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        label_encoders[col] = le
    return df, label_encoders

adultFrame_sample_cleaned, encoders = encode_categorical_columns(
    adultFrame_sample_cleaned
)
```

```python
# Step 3: Normalise numerical columns using Min-Max Scaling
scaler = MinMaxScaler()
adultFrame_sample_cleaned[numerical_columns] = scaler.fit_transform(
    adultFrame_sample_cleaned[numerical_columns]
)

adultFrame_sample_cleaned
```

The dataset contains missing data encoded in two forms: true NaN (from the CSV parser) and the string '?'. Both must be addressed; replace('?', pd.NA) followed by dropna() handles both in one pass. After cleaning, the sample reduces from 10,000 to 9,241 rows, meaning approximately 7.6% of sampled rows contained at least one missing or '?' value.

## P 2.3: Write a function to split the dataset in half column-wise and swap the first half and the second half. Apply the funtion to the dataset and print the result (3%)

```python
############# WRITE YOUR CODE IN THIS CELL (IF APPLICABLE)  ####################

def swap_column_halves(df):
    num_cols    = len(df.columns)
    mid         = num_cols // 2
    first_half  = df.iloc[:, :mid]
    second_half = df.iloc[:, mid:]
    swapped_df  = pd.concat([second_half, first_half], axis=1)
    return swapped_df

swapped_df = swap_column_halves(adultFrame_sample_cleaned)
print("Original:", list(adultFrame_sample_cleaned.columns))
print("Swapped :", list(swapped_df.columns))
swapped_df.head()
```
